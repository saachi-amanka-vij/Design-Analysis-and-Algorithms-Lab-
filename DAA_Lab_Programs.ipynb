## LAB 1.1 – Iterative Insertion Sort

```python
import random
import time
import matplotlib.pyplot as plt

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

arr = [5, 2, 4, 1, 3]
print("Sorted Array:", insertion_sort(arr))

sizes = [100, 200, 400, 800, 1200]
times = []
for n in sizes:
    data = [random.randint(1, 10000) for _ in range(n)]
    start = time.perf_counter()
    insertion_sort(data)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Insertion Sort (Iterative)")
plt.grid(True)
plt.show()
```

## LAB 1.2 – Recursive Insertion Sort

```python
import random
import time
import matplotlib.pyplot as plt
import sys

sys.setrecursionlimit(5000)

def insertion_sort(arr, n):
    if n <= 1:
        return
    insertion_sort(arr, n - 1)
    key = arr[n - 1]
    j = n - 2
    while j >= 0 and arr[j] > key:
        arr[j + 1] = arr[j]
        j -= 1
    arr[j + 1] = key

arr = [5, 2, 4, 1, 3]
insertion_sort(arr, len(arr))
print("Sorted Array:", arr)

sizes = [100, 200, 400, 800, 1200]
times = []
for n in sizes:
    data = [random.randint(1, 10000) for _ in range(n)]
    start = time.perf_counter()
    insertion_sort(data, len(data))
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Insertion Sort (Recursive)")
plt.grid(True)
plt.show()
```

## LAB 1.3 – Dutch National Flag Algorithm

```python
import random
import time
import matplotlib.pyplot as plt

def dutch_flag(arr):
    low = 0
    mid = 0
    high = len(arr) - 1
    while mid <= high:
        if arr[mid] == 0:
            arr[low], arr[mid] = arr[mid], arr[low]
            low += 1
            mid += 1
        elif arr[mid] == 1:
            mid += 1
        elif arr[mid] == 2:
            arr[mid], arr[high] = arr[high], arr[mid]
            high -= 1
        else:
            raise ValueError("Array must contain only 0, 1 and 2")
    return arr

arr = [2, 0, 2, 1, 1, 0]
print("Sorted Array:", dutch_flag(arr))

sizes = [100, 200, 400, 800, 1200]
times = []
for n in sizes:
    data = [random.randint(0, 2) for _ in range(n)]
    start = time.perf_counter()
    dutch_flag(data)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Dutch National Flag Algorithm")
plt.grid(True)
plt.show()
```

## LAB 1.4 – Majority Element (Boyer-Moore Voting)

```python
import random
import time
import matplotlib.pyplot as plt

def majority_element(arr):
    candidate = None
    count = 0
    for num in arr:
        if count == 0:
            candidate = num
        if num == candidate:
            count += 1
        else:
            count -= 1
    if arr.count(candidate) > len(arr) // 2:
        return candidate
    return -1

arr = [2, 2, 1, 2, 3, 2, 2]
print("Majority Element:", majority_element(arr))

sizes = [100, 200, 400, 800, 1200]
times = []
for n in sizes:
    data = [random.randint(1, 100) for _ in range(n)]
    data[:n // 2 + 1] = [999] * (n // 2 + 1)
    start = time.perf_counter()
    majority_element(data)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Majority Element")
plt.grid(True)
plt.show()
```

## LAB 2.1 – Union and Intersection of Two Sorted Arrays

```python
import random
import time
import matplotlib.pyplot as plt

def union_intersection(a, b):
    i = 0
    j = 0
    union = []
    intersection = []

    while i < len(a) and j < len(b):
        if a[i] == b[j]:
            if not union or union[-1] != a[i]:
                union.append(a[i])
            if not intersection or intersection[-1] != a[i]:
                intersection.append(a[i])
            i += 1
            j += 1
        elif a[i] < b[j]:
            if not union or union[-1] != a[i]:
                union.append(a[i])
            i += 1
        else:
            if not union or union[-1] != b[j]:
                union.append(b[j])
            j += 1

    while i < len(a):
        if not union or union[-1] != a[i]:
            union.append(a[i])
        i += 1

    while j < len(b):
        if not union or union[-1] != b[j]:
            union.append(b[j])
        j += 1

    return union, intersection

a = [1, 2, 3, 4, 5]
b = [2, 3, 4, 6, 7]
u, inter = union_intersection(a, b)
print("Union:", u)
print("Intersection:", inter)

sizes = [100, 200, 400, 800, 1200]
times = []
for n in sizes:
    a = sorted(random.sample(range(10000), n))
    b = sorted(random.sample(range(10000), n))
    start = time.perf_counter()
    union_intersection(a, b)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Union and Intersection")
plt.grid(True)
plt.show()
```

## LAB 2.2 – Search in a Rotated Sorted Array

```python
import random
import time
import matplotlib.pyplot as plt

def rotated_search(arr, target):
    low = 0
    high = len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        if arr[low] <= arr[mid]:
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1

arr = [4, 5, 6, 7, 0, 1, 2]
target = 0
print("Element Found at Index:", rotated_search(arr, target))

sizes = [100, 200, 400, 800, 1200, 2000]
times = []
for n in sizes:
    data = list(range(n))
    pivot = random.randint(0, n - 1)
    data = data[pivot:] + data[:pivot]
    target = n // 2
    start = time.perf_counter()
    rotated_search(data, target)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Search in Rotated Sorted Array")
plt.grid(True)
plt.show()
```

## LAB 2.3 – Count Inversions Using Merge Sort

```python
import random
import time
import matplotlib.pyplot as plt

def merge_count(arr):
    if len(arr) <= 1:
        return arr, 0

    mid = len(arr) // 2
    left, x = merge_count(arr[:mid])
    right, y = merge_count(arr[mid:])
    merged = []
    i = 0
    j = 0
    count = x + y

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            count += len(left) - i
            j += 1

    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged, count

def count_inversions(arr):
    return merge_count(arr)[1]

arr = [2, 4, 1, 3, 5]
print("Number of Inversions:", count_inversions(arr))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    data = [random.randint(1, 10000) for _ in range(n)]
    start = time.perf_counter()
    count_inversions(data)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Inversion Count Using Merge Sort")
plt.grid(True)
plt.show()
```

## LAB 3.1 – Matrix Multiplication (Brute Force)

```python
import random
import time
import matplotlib.pyplot as plt

def matrix_multiply(A, B):
    rows = len(A)
    cols = len(B[0])
    common = len(B)
    if len(A[0]) != common:
        raise ValueError("Invalid matrix dimensions")
    result = [[0 for _ in range(cols)] for _ in range(rows)]
    for i in range(rows):
        for j in range(cols):
            for k in range(common):
                result[i][j] += A[i][k] * B[k][j]
    return result

A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
print("Result:")
for row in matrix_multiply(A, B):
    print(row)

sizes = [10, 20, 30, 40, 50]
times = []
for n in sizes:
    A = [[random.randint(1, 10) for _ in range(n)] for _ in range(n)]
    B = [[random.randint(1, 10) for _ in range(n)] for _ in range(n)]
    start = time.perf_counter()
    matrix_multiply(A, B)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Matrix Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Brute Force Matrix Multiplication")
plt.grid(True)
plt.show()
```

## LAB 3.2 – Strassen's Matrix Multiplication

```python
import random
import time
import matplotlib.pyplot as plt

def add(A, B):
    n = len(A)
    return [[A[i][j] + B[i][j] for j in range(n)] for i in range(n)]

def subtract(A, B):
    n = len(A)
    return [[A[i][j] - B[i][j] for j in range(n)] for i in range(n)]

def strassen(A, B):
    n = len(A)
    if n == 1:
        return [[A[0][0] * B[0][0]]]

    mid = n // 2
    A11 = [row[:mid] for row in A[:mid]]
    A12 = [row[mid:] for row in A[:mid]]
    A21 = [row[:mid] for row in A[mid:]]
    A22 = [row[mid:] for row in A[mid:]]
    B11 = [row[:mid] for row in B[:mid]]
    B12 = [row[mid:] for row in B[:mid]]
    B21 = [row[:mid] for row in B[mid:]]
    B22 = [row[mid:] for row in B[mid:]]

    M1 = strassen(add(A11, A22), add(B11, B22))
    M2 = strassen(add(A21, A22), B11)
    M3 = strassen(A11, subtract(B12, B22))
    M4 = strassen(A22, subtract(B21, B11))
    M5 = strassen(add(A11, A12), B22)
    M6 = strassen(subtract(A21, A11), add(B11, B12))
    M7 = strassen(subtract(A12, A22), add(B21, B22))

    C11 = add(subtract(add(M1, M4), M5), M7)
    C12 = add(M3, M5)
    C21 = add(M2, M4)
    C22 = add(subtract(add(M1, M3), M2), M6)

    top = [C11[i] + C12[i] for i in range(mid)]
    bottom = [C21[i] + C22[i] for i in range(mid)]
    return top + bottom

def matrix_multiply_strassen(A, B):
    rows = len(A)
    common = len(B)
    cols = len(B[0])
    if len(A[0]) != common:
        raise ValueError("Invalid matrix dimensions")

    n = 1
    while n < max(rows, common, cols):
        n *= 2

    X = [[0] * n for _ in range(n)]
    Y = [[0] * n for _ in range(n)]

    for i in range(rows):
        for j in range(common):
            X[i][j] = A[i][j]
    for i in range(common):
        for j in range(cols):
            Y[i][j] = B[i][j]

    C = strassen(X, Y)
    return [row[:cols] for row in C[:rows]]

A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
print("Strassen Result:")
for row in matrix_multiply_strassen(A, B):
    print(row)

sizes = [2, 4, 8, 16, 32]
times = []
for n in sizes:
    A = [[random.randint(1, 10) for _ in range(n)] for _ in range(n)]
    B = [[random.randint(1, 10) for _ in range(n)] for _ in range(n)]
    start = time.perf_counter()
    matrix_multiply_strassen(A, B)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Matrix Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Strassen Matrix Multiplication")
plt.grid(True)
plt.show()
```

## LAB 3.3 – Book Allocation Problem

```python
import random
import time
import matplotlib.pyplot as plt

def allocate_books(pages, students):
    if students <= 0 or students > len(pages):
        return -1

    low = max(pages)
    high = sum(pages)
    answer = high

    while low <= high:
        mid = (low + high) // 2
        count = 1
        current = 0

        for page in pages:
            if current + page > mid:
                count += 1
                current = page
            else:
                current += page

        if count <= students:
            answer = mid
            high = mid - 1
        else:
            low = mid + 1

    return answer

pages = [12, 34, 67, 90]
students = 2
print("Minimum Maximum Pages:", allocate_books(pages, students))

sizes = [100, 200, 400, 800, 1200]
times = []
for n in sizes:
    data = [random.randint(1, 100) for _ in range(n)]
    start = time.perf_counter()
    allocate_books(data, n // 2)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Number of Books")
plt.ylabel("Execution Time (seconds)")
plt.title("Book Allocation Problem")
plt.grid(True)
plt.show()
```

## LAB 3.4 – Randomized Quick Sort

```python
import random
import time
import matplotlib.pyplot as plt

def randomized_quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = random.choice(arr)
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return randomized_quick_sort(left) + middle + randomized_quick_sort(right)

arr = [8, 3, 1, 7, 0, 10, 2]
print("Sorted Array:", randomized_quick_sort(arr))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    data = [random.randint(1, 10000) for _ in range(n)]
    start = time.perf_counter()
    randomized_quick_sort(data)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Randomized Quick Sort")
plt.grid(True)
plt.show()
```

## LAB 3.5 – Median of Two Sorted Arrays

```python
import random
import time
import matplotlib.pyplot as plt

def find_median(a, b):
    if len(a) > len(b):
        a, b = b, a

    m = len(a)
    n = len(b)

    if m + n == 0:
        return None

    low = 0
    high = m

    while low <= high:
        i = (low + high) // 2
        j = (m + n + 1) // 2 - i

        left_a = float("-inf") if i == 0 else a[i - 1]
        right_a = float("inf") if i == m else a[i]
        left_b = float("-inf") if j == 0 else b[j - 1]
        right_b = float("inf") if j == n else b[j]

        if left_a <= right_b and left_b <= right_a:
            if (m + n) % 2 == 0:
                return (max(left_a, left_b) + min(right_a, right_b)) / 2
            return max(left_a, left_b)
        elif left_a > right_b:
            high = i - 1
        else:
            low = i + 1

    raise ValueError("Both input arrays must be sorted")

a = [1, 3]
b = [2, 4]
print("Median:", find_median(a, b))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    a = sorted(random.sample(range(10000), n))
    b = sorted(random.sample(range(10000), n))
    start = time.perf_counter()
    find_median(a, b)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Median of Two Sorted Arrays")
plt.grid(True)
plt.show()
```

## LAB 4.1 – Maximum Product Subarray

```python
import random
import time
import matplotlib.pyplot as plt

def max_product(arr):
    if not arr:
        return 0

    maximum = arr[0]
    minimum = arr[0]
    result = arr[0]

    for i in range(1, len(arr)):
        if arr[i] < 0:
            maximum, minimum = minimum, maximum

        maximum = max(arr[i], maximum * arr[i])
        minimum = min(arr[i], minimum * arr[i])
        result = max(result, maximum)

    return result

arr = [2, 3, -2, 4]
print("Maximum Product:", max_product(arr))

sizes = [100, 200, 400, 800, 1200]
times = []
for n in sizes:
    data = [random.randint(-10, 10) for _ in range(n)]
    start = time.perf_counter()
    max_product(data)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Maximum Product Subarray")
plt.grid(True)
plt.show()
```

## LAB 4.2 – Kth Largest Element Using Quickselect

```python
import random
import time
import matplotlib.pyplot as plt

def kth_largest(arr, k):
    if k < 1 or k > len(arr):
        return -1

    arr = arr.copy()
    target = len(arr) - k
    low = 0
    high = len(arr) - 1

    while low <= high:
        pivot = arr[random.randint(low, high)]
        i = low
        j = low
        p = high

        while j <= p:
            if arr[j] < pivot:
                arr[i], arr[j] = arr[j], arr[i]
                i += 1
                j += 1
            elif arr[j] > pivot:
                arr[j], arr[p] = arr[p], arr[j]
                p -= 1
            else:
                j += 1

        if target < i:
            high = i - 1
        elif target > p:
            low = p + 1
        else:
            return pivot

    return -1

arr = [3, 2, 1, 5, 6, 4]
print("3rd Largest Element:", kth_largest(arr, 3))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    data = [random.randint(1, 10000) for _ in range(n)]
    start = time.perf_counter()
    kth_largest(data, n // 2)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Kth Largest Using Quickselect")
plt.grid(True)
plt.show()
```

## LAB 4.3 – Fibonacci Using Recursion and Dynamic Programming

```python
import time
import matplotlib.pyplot as plt

def fibonacci_recursive(n):
    if n <= 1:
        return n
    return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)

def fibonacci_dp(n):
    dp = [0] * (n + 1)
    if n >= 1:
        dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    return dp[n]

n = 8
print("Recursive Fibonacci:", fibonacci_recursive(n))
print("DP Fibonacci:", fibonacci_dp(n))

sizes = [5, 10, 15, 20, 25, 30]
recursive_times = []
dp_times = []
for n in sizes:
    start = time.perf_counter()
    fibonacci_recursive(n)
    recursive_times.append(time.perf_counter() - start)

    start = time.perf_counter()
    fibonacci_dp(n)
    dp_times.append(time.perf_counter() - start)

plt.plot(sizes, recursive_times, marker="o", label="Recursive")
plt.plot(sizes, dp_times, marker="s", label="Dynamic Programming")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Fibonacci Comparison")
plt.legend()
plt.grid(True)
plt.show()
```

## LAB 4.4 – Smallest Missing Positive Integer

```python
import random
import time
import matplotlib.pyplot as plt

def smallest_missing(arr):
    arr = arr.copy()
    n = len(arr)

    for i in range(n):
        while 1 <= arr[i] <= n and arr[arr[i] - 1] != arr[i]:
            j = arr[i] - 1
            arr[i], arr[j] = arr[j], arr[i]

    for i in range(n):
        if arr[i] != i + 1:
            return i + 1

    return n + 1

arr = [3, 4, -1, 1]
print("Smallest Missing Positive:", smallest_missing(arr))

sizes = [100, 200, 400, 800, 1200]
times = []
for n in sizes:
    data = [random.randint(-10, n) for _ in range(n)]
    start = time.perf_counter()
    smallest_missing(data)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Smallest Missing Positive")
plt.grid(True)
plt.show()
```

# LAB 5 – Heap-Based Algorithms

## LAB 5.1 – Heap Sort

```python
import random
import time
import matplotlib.pyplot as plt

def heapify(arr, n, i):
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left
    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)

def heap_sort(arr):
    arr = arr.copy()
    n = len(arr)

    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    for i in range(n - 1, 0, -1):
        arr[0], arr[i] = arr[i], arr[0]
        heapify(arr, i, 0)

    return arr

arr = [9, 4, 7, 1, 3, 6]
print("Sorted Array:", heap_sort(arr))

sizes = [100, 200, 400, 800, 1600, 3200]
times = []
for n in sizes:
    data = [random.randint(1, 10000) for _ in range(n)]
    start = time.perf_counter()
    heap_sort(data)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Heap Sort")
plt.grid(True)
plt.show()
```

## LAB 5.2 – Rearrange String With No Adjacent Equal Characters

```python
import random
import time
import matplotlib.pyplot as plt
import heapq
from collections import Counter

def rearrange_string(s):
    count = Counter(s)
    heap = [(-freq, char) for char, freq in count.items()]
    heapq.heapify(heap)

    result = []
    previous = None

    while heap:
        freq, char = heapq.heappop(heap)
        result.append(char)

        if previous is not None:
            heapq.heappush(heap, previous)

        freq += 1
        if freq < 0:
            previous = (freq, char)
        else:
            previous = None

    answer = "".join(result)
    if len(answer) != len(s):
        return "Not Possible"
    return answer

s = "aaabbc"
print("Rearranged String:", rearrange_string(s))

sizes = [100, 200, 400, 800, 1200]
times = []
for n in sizes:
    data = "".join(random.choice("abcdef") for _ in range(n))
    start = time.perf_counter()
    rearrange_string(data)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("String Length (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Rearrange String")
plt.grid(True)
plt.show()
```

## LAB 5.3 – Merge K Sorted Arrays Using a Heap

```python
import random
import time
import matplotlib.pyplot as plt
import heapq

def merge_k_arrays(arrays):
    heap = []
    for i in range(len(arrays)):
        if arrays[i]:
            heapq.heappush(heap, (arrays[i][0], i, 0))

    result = []
    while heap:
        value, array_no, index = heapq.heappop(heap)
        result.append(value)

        if index + 1 < len(arrays[array_no]):
            next_value = arrays[array_no][index + 1]
            heapq.heappush(heap, (next_value, array_no, index + 1))

    return result

arrays = [[1, 4, 7], [2, 5, 8], [3, 6, 9]]
print("Merged Array:", merge_k_arrays(arrays))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    arrays = []
    for i in range(5):
        data = sorted(random.randint(1, 10000) for _ in range(n))
        arrays.append(data)

    start = time.perf_counter()
    merge_k_arrays(arrays)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Elements in Each Array")
plt.ylabel("Execution Time (seconds)")
plt.title("Merge K Sorted Arrays")
plt.grid(True)
plt.show()
```

## LAB 5.4 – Kth Smallest Using Median of Medians

```python
import random
import time
import matplotlib.pyplot as plt

def kth_smallest(arr, k):
    if k < 1 or k > len(arr):
        return -1

    def select(a, k_index):
        if len(a) <= 5:
            return sorted(a)[k_index]

        groups = [a[i:i + 5] for i in range(0, len(a), 5)]
        medians = [sorted(group)[len(group) // 2] for group in groups]
        pivot = select(medians, len(medians) // 2)

        left = [x for x in a if x < pivot]
        equal = [x for x in a if x == pivot]
        right = [x for x in a if x > pivot]

        if k_index < len(left):
            return select(left, k_index)
        elif k_index < len(left) + len(equal):
            return pivot
        else:
            return select(right, k_index - len(left) - len(equal))

    return select(arr, k - 1)

arr = [7, 10, 4, 3, 20, 15]
print("3rd Smallest Element:", kth_smallest(arr, 3))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    data = [random.randint(1, 10000) for _ in range(n)]
    start = time.perf_counter()
    kth_smallest(data, n // 2)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Median of Medians")
plt.grid(True)
plt.show()
```

# LAB 6 – Linear-Time Sorting

## LAB 6.1 – Counting Sort

```python
import random
import time
import matplotlib.pyplot as plt

def counting_sort(arr):
    if len(arr) == 0:
        return []

    minimum = min(arr)
    maximum = max(arr)
    count = [0] * (maximum - minimum + 1)

    for num in arr:
        count[num - minimum] += 1

    result = []
    for i in range(len(count)):
        result.extend([i + minimum] * count[i])

    return result

arr = [4, 2, 2, 8, 3, 3, 1]
print("Sorted Array:", counting_sort(arr))

sizes = [100, 200, 400, 800, 1600, 3200]
times = []
for n in sizes:
    data = [random.randint(0, 1000) for _ in range(n)]
    start = time.perf_counter()
    counting_sort(data)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Counting Sort")
plt.grid(True)
plt.show()
```

## LAB 6.2 – Bucket Sort

```python
import random
import time
import matplotlib.pyplot as plt

def bucket_sort(arr):
    n = len(arr)
    if n == 0:
        return []

    minimum = min(arr)
    maximum = max(arr)

    if minimum == maximum:
        return arr.copy()

    buckets = [[] for _ in range(n)]

    for num in arr:
        index = int((num - minimum) / (maximum - minimum) * n)
        if index == n:
            index = n - 1
        buckets[index].append(num)

    result = []
    for bucket in buckets:
        bucket.sort()
        result.extend(bucket)

    return result

arr = [0.42, 0.32, 0.23, 0.52, 0.25, 0.47, 0.51]
print("Sorted Array:", bucket_sort(arr))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    data = [random.random() for _ in range(n)]
    start = time.perf_counter()
    bucket_sort(data)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Bucket Sort")
plt.grid(True)
plt.show()
```

## LAB 6.3 – Count Subarrays With Exactly K Distinct Elements

```python
import random
import time
import matplotlib.pyplot as plt

def subarrays_k_distinct(arr, k):
    if k <= 0:
        return 0

    def at_most(limit):
        count = {}
        left = 0
        result = 0

        for right in range(len(arr)):
            count[arr[right]] = count.get(arr[right], 0) + 1

            while len(count) > limit:
                count[arr[left]] -= 1
                if count[arr[left]] == 0:
                    del count[arr[left]]
                left += 1

            result += right - left + 1

        return result

    return at_most(k) - at_most(k - 1)

arr = [1, 2, 1, 2, 3]
print("Subarrays with Exactly 2 Distinct Elements:",
      subarrays_k_distinct(arr, 2))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    data = [random.randint(1, 10) for _ in range(n)]
    start = time.perf_counter()
    subarrays_k_distinct(data, 3)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Subarrays With K Distinct Elements")
plt.grid(True)
plt.show()
```

## LAB 6.4 – Minimize Sum After K Operations

```python
import random
import time
import matplotlib.pyplot as plt
import heapq

def minimize_sum(arr, k):
    heap = [-x for x in arr]
    heapq.heapify(heap)

    for _ in range(k):
        largest = -heapq.heappop(heap)
        new_value = (largest + 1) // 2
        heapq.heappush(heap, -new_value)

    return -sum(heap)

arr = [10, 20, 7]
k = 3
print("Minimum Sum:", minimize_sum(arr, k))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    data = [random.randint(1, 10000) for _ in range(n)]
    start = time.perf_counter()
    minimize_sum(data, 100)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Minimize Sum After K Operations")
plt.grid(True)
plt.show()
```

# LAB 7 – Hashing Techniques

## LAB 7.1 – Hash Table Using Separate Chaining

```python
import random
import time
import matplotlib.pyplot as plt

class HashTable:
    def __init__(self, size):
        self.size = size
        self.table = [[] for _ in range(size)]

    def hash_function(self, key):
        return key % self.size

    def insert(self, key):
        index = self.hash_function(key)
        if key not in self.table[index]:
            self.table[index].append(key)

    def search(self, key):
        index = self.hash_function(key)
        return key in self.table[index]

    def display(self):
        print(self.table)

ht = HashTable(10)
for x in [10, 20, 15, 7]:
    ht.insert(x)

print("Hash Table:")
ht.display()
print("Search 15:", ht.search(15))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    ht = HashTable(n)
    for i in range(n):
        ht.insert(i)

    start = time.perf_counter()
    for _ in range(1000):
        ht.search(n - 1)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Number of Elements")
plt.ylabel("Execution Time (seconds)")
plt.title("Hashing Using Separate Chaining")
plt.grid(True)
plt.show()
```

## LAB 7.2 – Linear Probing in Hashing

```python
import random
import time
import matplotlib.pyplot as plt

class LinearProbing:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    def hash_function(self, key):
        return key % self.size

    def insert(self, key):
        index = self.hash_function(key)

        for i in range(self.size):
            new_index = (index + i) % self.size

            if self.table[new_index] == key:
                return

            if self.table[new_index] is None:
                self.table[new_index] = key
                return

        raise ValueError("Hash Table is Full")

    def search(self, key):
        index = self.hash_function(key)

        for i in range(self.size):
            new_index = (index + i) % self.size

            if self.table[new_index] is None:
                return False

            if self.table[new_index] == key:
                return True

        return False

    def display(self):
        print(self.table)

ht = LinearProbing(10)
for x in [10, 20, 15, 7]:
    ht.insert(x)

print("Hash Table:")
ht.display()
print("Search 20:", ht.search(20))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    ht = LinearProbing(n * 2)
    for i in range(n):
        ht.insert(i)

    start = time.perf_counter()
    for _ in range(1000):
        ht.search(n - 1)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Number of Elements")
plt.ylabel("Execution Time (seconds)")
plt.title("Linear Probing")
plt.grid(True)
plt.show()
```

## LAB 7.3 – Minimum Window Substring Using Hashing

```python
import random
import time
import matplotlib.pyplot as plt

def minimum_window(s, t):
    if not s or not t:
        return ""

    need = {}
    for ch in t:
        need[ch] = need.get(ch, 0) + 1

    left = 0
    missing = len(t)
    start = 0
    min_length = float("inf")

    for right, ch in enumerate(s):
        if need.get(ch, 0) > 0:
            missing -= 1
        need[ch] = need.get(ch, 0) - 1

        while missing == 0:
            if right - left + 1 < min_length:
                min_length = right - left + 1
                start = left

            old = s[left]
            need[old] = need.get(old, 0) + 1

            if need[old] > 0:
                missing += 1

            left += 1

    if min_length == float("inf"):
        return ""

    return s[start:start + min_length]

s = "ADOBECODEBANC"
t = "ABC"
print("Minimum Window:", minimum_window(s, t))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    s = "".join(random.choice("ABCDEF") for _ in range(n))
    start = time.perf_counter()
    minimum_window(s, "ABC")
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("String Length (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Minimum Window Substring")
plt.grid(True)
plt.show()
```

## LAB 8.1 – Coin Change Using Dynamic Programming

```python
import time
import matplotlib.pyplot as plt

def coin_change(coins, amount):
    dp = [float("inf")] * (amount + 1)
    dp[0] = 0

    for i in range(1, amount + 1):
        for coin in coins:
            if coin <= i:
                dp[i] = min(dp[i], dp[i - coin] + 1)

    if dp[amount] == float("inf"):
        return -1

    return dp[amount]

coins = [1, 2, 5]
amount = 11
print("Minimum Coins Required:", coin_change(coins, amount))

sizes = [100, 200, 400, 800, 1200, 1600]
times = []
for n in sizes:
    start = time.perf_counter()
    coin_change([1, 3, 4, 5], n)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Amount")
plt.ylabel("Execution Time (seconds)")
plt.title("Coin Change Using DP")
plt.grid(True)
plt.show()
```

## LAB 8.3 – Maximize Minimum Distance (Aggressive Cows)

```python
import random
import time
import matplotlib.pyplot as plt

def maximize_min_distance(arr, cows):
    if cows <= 0 or cows > len(arr):
        return -1
    if cows == 1:
        return 0

    arr = sorted(arr)
    low = 1
    high = arr[-1] - arr[0]
    answer = 0

    while low <= high:
        mid = (low + high) // 2
        count = 1
        last_position = arr[0]

        for i in range(1, len(arr)):
            if arr[i] - last_position >= mid:
                count += 1
                last_position = arr[i]

        if count >= cows:
            answer = mid
            low = mid + 1
        else:
            high = mid - 1

    return answer

arr = [1, 2, 8, 4, 9]
cows = 3
print("Maximum Minimum Distance:", maximize_min_distance(arr, cows))

sizes = [100, 200, 400, 800, 1200]
times = []
for n in sizes:
    data = random.sample(range(n * 10), n)
    start = time.perf_counter()
    maximize_min_distance(data, n // 2)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Number of Positions")
plt.ylabel("Execution Time (seconds)")
plt.title("Maximum Minimum Distance")
plt.grid(True)
plt.show()
```

## LAB 9.1 – Construct a BST Using Insert Function

```python
import random
import time
import matplotlib.pyplot as plt
import sys

sys.setrecursionlimit(10000)

class Node:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None

def insert(root, key):
    if root is None:
        return Node(key)
    if key < root.key:
        root.left = insert(root.left, key)
    elif key > root.key:
        root.right = insert(root.right, key)
    return root

def inorder(root):
    if root is None:
        return []
    return inorder(root.left) + [root.key] + inorder(root.right)

root = None
values = [50, 30, 70, 20, 40, 60, 80]
for value in values:
    root = insert(root, value)

print("BST Inorder Traversal:", inorder(root))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    data = random.sample(range(n * 10), n)
    start = time.perf_counter()
    root = None
    for x in data:
        root = insert(root, x)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Number of Nodes")
plt.ylabel("Execution Time (seconds)")
plt.title("Binary Search Tree Insertion")
plt.grid(True)
plt.show()
```

## LAB 9.2 – Lowest Common Ancestor in a BST

```python
import random
import time
import matplotlib.pyplot as plt
import sys

sys.setrecursionlimit(10000)

class Node:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None

def insert(root, key):
    if root is None:
        return Node(key)
    if key < root.key:
        root.left = insert(root.left, key)
    elif key > root.key:
        root.right = insert(root.right, key)
    return root

def search(root, key):
    while root:
        if root.key == key:
            return True
        elif key < root.key:
            root = root.left
        else:
            root = root.right
    return False

def lowest_common_ancestor(root, n1, n2):
    if not search(root, n1) or not search(root, n2):
        return None

    while root:
        if n1 < root.key and n2 < root.key:
            root = root.left
        elif n1 > root.key and n2 > root.key:
            root = root.right
        else:
            return root.key
    return None

root = None
for x in [50, 30, 70, 20, 40, 60, 80]:
    root = insert(root, x)

print("LCA of 20 and 40:", lowest_common_ancestor(root, 20, 40))
print("LCA of 20 and 80:", lowest_common_ancestor(root, 20, 80))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    data = random.sample(range(n * 10), n)
    root = None
    for x in data:
        root = insert(root, x)

    start = time.perf_counter()
    lowest_common_ancestor(root, data[0], data[-1])
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Number of Nodes")
plt.ylabel("Execution Time (seconds)")
plt.title("Lowest Common Ancestor in BST")
plt.grid(True)
plt.show()
```

# LAB 10 – Minimum Spanning Tree and DP on Sequences

## LAB 10.1 – Prim's Minimum Spanning Tree Algorithm

```python
import random
import time
import matplotlib.pyplot as plt
import heapq

def prim_mst(graph, start):
    visited = set()
    heap = [(0, start, None)]
    mst = []
    total_weight = 0

    while heap:
        weight, node, parent = heapq.heappop(heap)
        if node in visited:
            continue

        visited.add(node)

        if parent is not None:
            mst.append((parent, node, weight))
            total_weight += weight

        for neighbor, w in graph[node]:
            if neighbor not in visited:
                heapq.heappush(heap, (w, neighbor, node))

    if len(visited) != len(graph):
        return [], -1

    return mst, total_weight

graph = {
    0: [(1, 2), (3, 6)],
    1: [(0, 2), (2, 3), (3, 8), (4, 5)],
    2: [(1, 3), (4, 7)],
    3: [(0, 6), (1, 8), (4, 9)],
    4: [(1, 5), (2, 7), (3, 9)]
}

mst, weight = prim_mst(graph, 0)
print("MST Edges:", mst)
print("Total MST Weight:", weight)

sizes = [50, 100, 200, 400, 800]
times = []
for n in sizes:
    graph = {i: [] for i in range(n)}

    # Ensure graph is connected
    for i in range(n - 1):
        w = random.randint(1, 20)
        graph[i].append((i + 1, w))
        graph[i + 1].append((i, w))

    # Add extra edges
    for _ in range(2 * n):
        u = random.randint(0, n - 1)
        v = random.randint(0, n - 1)
        if u != v:
            w = random.randint(1, 20)
            graph[u].append((v, w))
            graph[v].append((u, w))

    start = time.perf_counter()
    prim_mst(graph, 0)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Number of Vertices")
plt.ylabel("Execution Time (seconds)")
plt.title("Prim's Minimum Spanning Tree")
plt.grid(True)
plt.show()
```

## LAB 10.2 – Count Nodes in a Complete Binary Tree

```python
import random
import time
import matplotlib.pyplot as plt

class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

def count_nodes(root):
    if root is None:
        return 0

    left_height = 0
    right_height = 0
    left = root
    right = root

    while left:
        left_height += 1
        left = left.left

    while right:
        right_height += 1
        right = right.right

    if left_height == right_height:
        return (2 ** left_height) - 1

    return 1 + count_nodes(root.left) + count_nodes(root.right)

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
root.right.left = Node(6)

print("Total Number of Nodes:", count_nodes(root))

sizes = [100, 200, 400, 800, 1600]
times = []
for n in sizes:
    nodes = [Node(i) for i in range(n)]
    for i in range(n):
        left = 2 * i + 1
        right = 2 * i + 2
        if left < n:
            nodes[i].left = nodes[left]
        if right < n:
            nodes[i].right = nodes[right]

    start = time.perf_counter()
    count_nodes(nodes[0])
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Number of Nodes")
plt.ylabel("Execution Time (seconds)")
plt.title("Count Nodes in Complete Binary Tree")
plt.grid(True)
plt.show()
```

## LAB 10.3 – Longest Palindromic Subsequence Using DP

```python
import random
import time
import matplotlib.pyplot as plt

def longest_palindrome(s):
    n = len(s)
    if n == 0:
        return 0

    dp = [[0] * n for _ in range(n)]

    for i in range(n - 1, -1, -1):
        dp[i][i] = 1
        for j in range(i + 1, n):
            if s[i] == s[j]:
                dp[i][j] = 2 + (dp[i + 1][j - 1] if j > i + 1 else 0)
            else:
                dp[i][j] = max(dp[i + 1][j], dp[i][j - 1])

    return dp[0][n - 1]

s = "BBABCBCAB"
print("Length of Longest Palindromic Subsequence:", longest_palindrome(s))

sizes = [50, 100, 150, 200, 250, 300]
times = []
for n in sizes:
    s = "".join(random.choice("ABC") for _ in range(n))
    start = time.perf_counter()
    longest_palindrome(s)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("String Length (n)")
plt.ylabel("Execution Time (seconds)")
plt.title("Longest Palindromic Subsequence")
plt.grid(True)
plt.show()
```

# LAB 11 – Single Source Shortest Path (Greedy)

## LAB 11.1 – Dijkstra's Shortest Path Algorithm

```python
import random
import time
import matplotlib.pyplot as plt
import heapq

def dijkstra(graph, source):
    distance = {node: float("inf") for node in graph}
    distance[source] = 0
    heap = [(0, source)]

    while heap:
        dist, node = heapq.heappop(heap)
        if dist > distance[node]:
            continue

        for neighbor, weight in graph[node]:
            new_distance = dist + weight
            if new_distance < distance[neighbor]:
                distance[neighbor] = new_distance
                heapq.heappush(heap, (new_distance, neighbor))

    return distance

graph = {
    0: [(1, 4), (2, 1)],
    1: [(3, 1)],
    2: [(1, 2), (3, 5)],
    3: [(4, 3)],
    4: []
}

print("Shortest Distances:", dijkstra(graph, 0))

sizes = [50, 100, 200, 400, 800]
times = []
for n in sizes:
    graph = {i: [] for i in range(n)}

    # Ensure every node is reachable from node 0
    for i in range(n - 1):
        w = random.randint(1, 20)
        graph[i].append((i + 1, w))

    # Add extra directed edges
    for _ in range(2 * n):
        u = random.randint(0, n - 1)
        v = random.randint(0, n - 1)
        if u != v:
            w = random.randint(1, 20)
            graph[u].append((v, w))

    start = time.perf_counter()
    dijkstra(graph, 0)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Number of Vertices")
plt.ylabel("Execution Time (seconds)")
plt.title("Dijkstra's Shortest Path Algorithm")
plt.grid(True)
plt.show()
```

## LAB 11.2 – Stock Buy and Sell With Transaction Fee

```python
import random
import time
import matplotlib.pyplot as plt

def max_profit(prices, fee):
    if len(prices) == 0:
        return 0

    cash = 0
    hold = -prices[0]

    for price in prices[1:]:
        old_cash = cash
        cash = max(cash, hold + price - fee)
        hold = max(hold, old_cash - price)

    return cash

prices = [1, 3, 2, 8, 4, 9]
fee = 2
print("Maximum Profit:", max_profit(prices, fee))

sizes = [100, 200, 400, 800, 1600, 3200]
times = []
for n in sizes:
    data = [random.randint(1, 1000) for _ in range(n)]
    start = time.perf_counter()
    max_profit(data, 2)
    times.append(time.perf_counter() - start)

plt.plot(sizes, times, marker="o")
plt.xlabel("Number of Days")
plt.ylabel("Execution Time (seconds)")
plt.title("Stock Buy and Sell With Transaction Fee")
plt.grid(True)
plt.show()
```
